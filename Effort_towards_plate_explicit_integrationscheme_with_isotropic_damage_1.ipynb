```python
def props_update(mat_params, i,l,j,k, pois):
    
    return (mat_params['E'+str(i+1)+str(l+1)+str(j+1)+str(k+1)]*pois)/((1 + pois)*(1 - 2*pois)), 0.5*mat_params['E'+str(i+1)+str(l+1)+str(j+1)+str(k+1)]/(1 + pois)


def tens_to_voigt(tens):
    
    stress    = np.zeros(shape=(6,))
    
    stress[0] = tens[0][0]
    stress[1] = tens[1][1]
    stress[2] = tens[2][2]
    stress[3] = tens[1][2]
    stress[4] = tens[0][2]
    stress[5] = tens[0][1]
    
    return stress


def voigt_to_tens(stress):
    
    tens = np.zeros(shape=(3,3))
    
    
    tens[0][0] = stress[0]
    tens[1][1] = stress[1]
    tens[2][2] = stress[2]
    tens[0][1] = stress[5]
    tens[0][2] = stress[4]
    tens[1][0] = stress[5]
    tens[1][2] = stress[3]
    tens[2][0] = stress[4]
    tens[2][1] = stress[3]
    
    
    return tens


def split(tens):
   
    I      = np.eye(3,3)
    
    aa     = voigt_to_tens(tens)
    
    shydro = (1/3*(aa.trace()))
    
    return aa - shydro*I, shydro*I
```

```python
def combine_pdfs(folder_path, output_filename):
    """
    Kombiniert alle PDFs in einem Ordner zu einer einzigen PDF-Datei.

    :param folder_path: Pfad zum Ordner, der die PDF-Dateien enthält.
    :param output_filename: Name der kombinierten PDF-Ausgabedatei.
    """
    merger = PdfMerger()

    # Alle Dateien im Ordner durchgehen
    for filename in sorted(os.listdir(folder_path)):
        if filename.endswith('.pdf'):
            pdf_path = os.path.join(folder_path, filename)
            print(f"Füge hinzu: {pdf_path}")
            merger.append(pdf_path)

    # Kombinierte PDF speichern
    output_path = os.path.join(folder_path, output_filename)
    merger.write(output_path)
    merger.close()
    print(f"Kombinierte PDF gespeichert unter: {output_path}")
```

```python
def extract_pdf_page(input_file, page_number, output_file):
    try:
        # PDF-Datei lesen
        reader = PdfReader(input_file)
        
        # Prüfen, ob die Seite existiert
        if page_number < 1 or page_number > len(reader.pages):
            print(f"Die Datei hat nur {len(reader.pages)} Seiten. Bitte eine gültige Seite angeben.")
            return
        
        # PDF-Schreiber erstellen
        writer = PdfWriter()
        
        # Bestimmte Seite extrahieren
        page = reader.pages[page_number - 1]  # Seiten beginnen bei Index 0
        writer.add_page(page)
        
        # Neue Datei speichern
        with open(output_file, "wb") as output_pdf:
            writer.write(output_pdf)
        
        print(f"Seite {page_number} wurde erfolgreich in '{output_file}' gespeichert.")
    
    except Exception as e:
        print(f"Fehler: {e}")

# Beispielaufruf
input_pdf = r"C:\Users\Tandale\sciebo\olddata\Einbuergerung\Zeugnisse\Einbuergerungsantrag_08-24.pdf"  # Ersetze mit dem Pfad zu deiner PDF-Datei
page_to_extract = 16      # Ersetze mit der gewünschten Seitennummer
output_pdf = r"C:\Users\Tandale\sciebo\olddata\Einbuergerung\Zeugnisse\Arbeitgeber_Bescheinigung.pdf" # Ersetze mit dem Namen der Ausgabe-PDF

```

```python
extract_pdf_page(input_pdf, page_to_extract, output_pdf)
```
